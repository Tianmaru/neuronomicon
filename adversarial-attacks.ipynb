# Adversarial Attacks (WIP)

Szegedy et al.: [Intriguing Properties of Neural Networks](https://doi.org/10.48550/arXiv.1312.6199) (2013)

- outcome 1: apparently, it is rather the embedded space than individual high-level neurons, that holds semantic information
- random linear combinations of high-level neurons appear to have a similar "meaning" as individual high-level neurons
- outcome 2: neural networks are fairly discontinous - different prediction for slightly altered input
- introduced the terms *adversarial sample* and *adversarial attack*

Minimize $loss_f (x + r,l) + c|r|$ subject to $x+r \in [0,1]^m$

- $x$: original image
- $l \neq f(x)$: target label of the adversarial attack, direction of the perturbation
- $r$: perturbation noise that's added to the image
- $c$: coefficient found via line-search that adds a penalty to find the closest adversarial image $x+r$ with $f(x+r)=l$

Shan et al.: [Prompt-Specific Poisnoning Attacks on Text-to-Image Generative Models](https://doi.org/10.48550/arXiv.2310.13828) (2023)

- Data Poisoning Attack: Injection of manipulated samples into training set of a machine learning model to cause unexpected behaviour
- a successfull data poisoning attack requires an amount of poisoned examples corresponding to 5-20% of clean training examples 
- Text-to-Image Models usually trained on large data sets, therefore assumed to be resistant to data poisoning
- however, paper shows that Text-to-Image Models are vulnerable due to *Concept Sparsity*
- large training data set, but only a small amount of samples per keyword/concept
- proposes *Nightshade* for optimized *prompt-specific poisoning attacks*
- create poisoned images for a specific concept using guided perturbation
- possible defense strategy for content creators to protect their copyright
